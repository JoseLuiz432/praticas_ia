# Anonimização de Dados Usando IA Generativa

Nesta aula prática, vamos explorar como técnicas de IA generativa podem ser utilizadas para anonimização de dados. Vamos usar um conjunto de documentos fictícios em formato de conversação com um atendente virtual, contendo dados sensíveis, e aplicar técnicas para proteger a privacidade dos indivíduos, substituindo dados sensíveis por informações geradas pela IA.

## Objetivos
- Entender os conceitos básicos de anonimização de dados.
- Utilizar modelos de linguagem para gerar substituições realistas de dados sensíveis.
- Avaliar os resultados obtidos com IA generativa.

```python
from openai import OpenAI
import dotenv
import os

dotenv.load_dotenv()

token = os.getenv("OPENAI_KEY")
client = OpenAI(api_key=token)
client
```

```python
assistente = client.beta.assistants.create(
  name="Especialista em anonimização de dados",
  instructions="Você é um especialista em anonimização de dados e foi contratado para revisar conversas entre atendentes e clientes."
     + "Seu objetivo é identificar e anonimizar informações sensíveis, como nomes, emails, telefones, endereços e CPFs. "
     + "Certifique-se de manter o contexto da conversa e garantir que os dados sensíveis sejam protegidos.",
  tools=[{"type": "code_interpreter"}],
  model="gpt-3.5-turbo",
)

assistente
```

```python
thread = client.beta.threads.create()
thread.to_dict()
```

```python
for index, row in df.iterrows():
    document = row['document']
    message = client.beta.threads.messages.create(
        thread_id=thread.id,
        role="user",
        content=f"Anonimize os dados sensíveis na seguinte conversa:\n\n Retornanr somente a conversa e nada mais{document}"
    )
    run = client.beta.threads.runs.create_and_poll(
        thread_id=thread.id,
        assistant_id=assistente.id
    )
```

```python
from IPython.display import display, Markdown

if run.status == "completed":
    messages = client.beta.threads.messages.list(thread_id=thread.id)

    for msg in reversed(messages.data):
        display(Markdown(f"**{msg.role}**"))
        display(Markdown(f"{msg.content[0].text.value}"))
else:
    print(run.status)
```
